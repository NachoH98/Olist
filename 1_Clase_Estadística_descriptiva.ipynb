```python
# Descargar datos financieros de Yahoo Finance
# Armamos una función para descargar datos de Yahoo Finance
def descargar_datos(ticker, inicio, fin):
    """
    Descarga datos históricos de Yahoo Finance
    """
    datos = yf.download(ticker, start=inicio, end=fin)
    return datos

# Ejemplo: Datos de Apple (AAPL) y del S&P 500 (^GSPC) como variable explicativa
inicio = '2023-01-01'
fin = '2025-12-31'

# Descargar datos
print("Descargando datos de AAPL y S&P 500...")
aapl = descargar_datos('AAPL', inicio, fin)
sp500 = descargar_datos('^GSPC', inicio, fin)

# Prepararamos los datos

datos = pd.DataFrame({
    'AAPL_Close': aapl['Close'].squeeze(),
    'SP500_Close': sp500['Close'].squeeze()
}, index=aapl.index).dropna()

# Calculamos los retornos diarios y lo agregamos al dataframe
datos['AAPL_Retorno'] = datos['AAPL_Close'].pct_change() * 100
datos['SP500_Retorno'] = datos['SP500_Close'].pct_change() * 100
datos = datos.dropna()

datos #visualizamos el dataframe
```

```python
datos.describe().transpose()
```

```python
# Otras operaciones son:
print(datos['AAPL_Retorno'].min()) # Minimo
print(datos['AAPL_Retorno'].max()) # Maximo
print(datos['AAPL_Retorno'].count()) # Cantidad
print(datos['AAPL_Retorno'].idxmax()) # El Ìndice del valor máximo
print(datos['AAPL_Retorno'].idxmin()) # El Ìndice del valor mínimo
print(datos['AAPL_Retorno'].quantile([.25,.5,.75])) # Los quantiles
print(datos['AAPL_Retorno'].skew()) # Asimetria
print(datos['AAPL_Retorno'].kurtosis()) # Kurtosis
```

```python
std_aapl_retorno = datos['AAPL_Retorno'].std()
mean_aapl_retorno = datos['AAPL_Retorno'].mean()

# Calculamos el Coeficiente de Variabilidad (CV)
# CV = (Desviación Estándar / Media) * 100
# Multiplicamos por 100 para expresarlo en porcentaje
if mean_aapl_retorno != 0:
    cv_aapl_retorno = (std_aapl_retorno / mean_aapl_retorno) * 100
    print(f"La desviación estándar de los retornos es: {std_aapl_retorno:.2f}")
    print(f"La media de los retornos es: {mean_aapl_retorno:.2f}")
    print(f"El coeficiente de variabilidad de los retornos es: {cv_aapl_retorno:.2f}%")
else:
    print("La media de los retornos es cero, el coeficiente de variabilidad no puede ser calculado.")
```

```python
std_sp_retorno = datos['SP500_Retorno'].std()
mean_sp_retorno = datos['SP500_Retorno'].mean()

# Calculamos el Coeficiente de Variabilidad (CV)
# CV = (Desviación Estándar / Media) * 100
# Multiplicamos por 100 para expresarlo en porcentaje
if mean_sp_retorno != 0:
    cv_sp_retorno = (std_sp_retorno / mean_sp_retorno) * 100
    print(f"La desviación estándar de los retornos es: {std_sp_retorno:.2f}")
    print(f"La media de los retornos es: {mean_sp_retorno:.2f}")
    print(f"El coeficiente de variabilidad de los retornos es: {cv_sp_retorno:.2f}%")
else:
    print("La media de los retornos es cero, el coeficiente de variabilidad no puede ser calculado.")
```

```python
# Visualizamos la relación entre los retornos
plt.figure(figsize=(10, 6))
sb.scatterplot(x='SP500_Retorno', y='AAPL_Retorno', data=datos)
plt.title('Relación entre Retornos de AAPL y S&P 500')
plt.xlabel('Retorno Diario S&P 500 (%)')
plt.ylabel('Retorno Diario AAPL (%)')
plt.grid(True)
plt.show()
```

El dataset Carseats contiene datos simulados sobre las ventas de asientos para niños en 400 tiendas diferentes. Este conjunto de datos incluye las siguientes variables:

Price: Precio que la empresa cobra por los asientos para niños en cada sitio.

US: Indicador de si la tienda está en los EE. UU. o no.

```python
# ejemplo función y concatenado

def obtener_Z(df, columna):
    """Obtiene el valor Z de cada elemento de una columna determinada
    de un archivo seleccionado"""
    df2 = pd.DataFrame([])
    df2[f"Z_{columna}"] =  (df[columna] - df[columna].mean()) / df[columna].std()
    return df2
```

```python
obtener_Z(datos, "Price")
```

```python
obtener_Z(datos, "CompPrice")
```

```python
#Armamos un dataframe usando la función y concatenando

Z_Price = obtener_Z(datos, "Price")
Z_CompPrice=obtener_Z(datos, "CompPrice")
archivo_3=pd.concat([Z_Price, Z_CompPrice ], axis=1)
archivo_3
```
