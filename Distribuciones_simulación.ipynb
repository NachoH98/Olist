```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sb
```

```python
#Se puede generar y trabajar con distribuciones de probabilidad.

## Distribución Normal

mu = 0 #definimos la media
sigma = 1 #definimos el desvío
#ahora,generamos 10000 números con media 0 y desvío 1
dist_norm = np.random.normal(mu, sigma, 10000)

#graficamos una normal 0-1 con 10000 valores simulados
sb.displot(dist_norm, kind = 'kde')
plt.title('Distribución Normal 0 - 1')
plt.show()
```

```python
#iteramos para una lista de n cantidad de variables
n_muestras = [10, 1000, 10000, 50000]
mu = 0
sigma = 1

for n in n_muestras:

    norm_dist = np.random.normal(mu, sigma, n)
    print("Para un n de: {} Media: {} Desvío: {}".format(n, norm_dist.mean(), norm_dist.std()))

    sb.kdeplot(norm_dist, linewidth = 2, label = n)

plt.legend(prop={'size': 8}, title = 'Cantidad de muestras')
plt.title('Distribuciónes Normales 0 -1')
plt.show()
```

```python
#Tomamos una lista con diferentes valores de mu y sigma
parametros = [(1,2.4),(4,1.5),(5,5)]
for mu, sigma in parametros:
    sb.kdeplot(np.random.normal(mu, sigma, 10000),
                 linewidth = 2, label = (mu,sigma))


plt.legend(prop={'size': 8}, title = 'Parámetros')
plt.title('Distribuciónes Normales con diferentes parámetros')
plt.show()
```

```python
## Distribución Chi Cuadrado
#Simulamos 10000 valores de una chi cuadrado para diferentes grados de libertad

for gl in [1,2,5,10,30,50]:
    sb.kdeplot(np.random.chisquare(gl,10000),
                 linewidth = 2, label = gl)

plt.legend(prop={'size': 8}, title = 'Grados de libertad')
plt.title('Distribución Chi Cuadrado')
plt.show()
```

```python
## Distribución Exponencial
#Simulamos una distribución exponencial para diferentes valores de beta

for beta in [1.5,2,5,10,20]:
    sb.kdeplot(np.random.exponential(beta,10000),
                 linewidth= 2, label = beta)

plt.legend(prop={'size': 8}, title = 'Beta')
plt.title('Distribución Exponencial')
plt.show()
```

```python
#Simulamos una distribución uniforme para diferentes rangos.
for low, high in [(0, 1), (5, 10), (-2, 2)]:
    sb.kdeplot(np.random.uniform(low, high, 50000),
                 linewidth=2, label=f'[{low}, {high}]')

plt.legend(prop={'size': 8}, title='Rangos')
plt.title('Distribución Uniforme')
plt.show()
```

```python
#Simulamos una distribución Gamma para diferentes parámetros de forma (shape) y escala (scale).
for shape, scale in [(1, 2), (2, 2), (5, 1)]:
    sb.kdeplot(np.random.gamma(shape, scale, 10000),
                 linewidth=2, label=f'k={shape}, theta={scale}')

plt.legend(prop={'size': 8}, title='Parámetros')
plt.title('Distribución Gamma')
plt.show()
```

```python
#Simulamos una distribución Binomial para diferentes número de (n) y (p).

for n_trials, p_success in [(10, 0.5), (20, 0.3), (5, 0.8)]:
    sb.histplot(np.random.binomial(n_trials, p_success, 10000),
                 stat='density', kde=False, label=f'n={n_trials}, p={p_success}', alpha=0.5)

plt.legend(prop={'size': 8}, title='Parámetros')
plt.title('Distribución Binomial')
plt.show()
```

```python
#Simulamos una distribución de Poisson para diferentes lambda.
for lam in [1, 5, 10]:
    sb.histplot(np.random.poisson(lam, 10000),
                 stat='density', kde=False, label=f'lambda={lam}', alpha=0.5)

plt.legend(prop={'size': 8}, title='Parámetro')
plt.title('Distribución de Poisson')
plt.show()
```
